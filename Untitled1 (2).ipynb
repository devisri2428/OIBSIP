```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

# Seaborn style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Reproducible results
np.random.seed(42)

# Number of sales transactions
n = 1000

# -----------------------------
# Basic customer information
# -----------------------------

customer_ids = [f"CUST{str(i).zfill(4)}" for i in range(1, 301)]

genders = np.random.choice(
    ["Male", "Female"],
    size=n,
    p=[0.48, 0.52]
)

ages = np.random.randint(18, 65, size=n)

# Create age groups
def age_group(age):
    if age <= 25:
        return "18-25"
    elif age <= 35:
        return "26-35"
    elif age <= 45:
        return "36-45"
    elif age <= 55:
        return "46-55"
    else:
        return "56-65"

age_groups = [age_group(age) for age in ages]

# -----------------------------
# Product information
# -----------------------------

products = [
    "Laptop",
    "Smartphone",
    "Headphones",
    "Keyboard",
    "Mouse",
    "Monitor",
    "Tablet",
    "Smart Watch",
    "Printer",
    "Backpack",
    "Office Chair",
    "Desk Lamp",
    "Coffee Maker",
    "Blender",
    "Running Shoes"
]

categories = {
    "Laptop": "Electronics",
    "Smartphone": "Electronics",
    "Headphones": "Electronics",
    "Keyboard": "Electronics",
    "Mouse": "Electronics",
    "Monitor": "Electronics",
    "Tablet": "Electronics",
    "Smart Watch": "Electronics",
    "Printer": "Electronics",
    "Backpack": "Accessories",
    "Office Chair": "Furniture",
    "Desk Lamp": "Furniture",
    "Coffee Maker": "Home Appliances",
    "Blender": "Home Appliances",
    "Running Shoes": "Sports"
}

prices = {
    "Laptop": 65000,
    "Smartphone": 35000,
    "Headphones": 2500,
    "Keyboard": 1800,
    "Mouse": 900,
    "Monitor": 12000,
    "Tablet": 22000,
    "Smart Watch": 7000,
    "Printer": 11000,
    "Backpack": 2500,
    "Office Chair": 8500,
    "Desk Lamp": 1500,
    "Coffee Maker": 6000,
    "Blender": 3500,
    "Running Shoes": 4500
}

# Product selection probabilities
product_probabilities = [
    0.07,  # Laptop
    0.10,  # Smartphone
    0.09,  # Headphones
    0.07,  # Keyboard
    0.08,  # Mouse
    0.07,  # Monitor
    0.06,  # Tablet
    0.07,  # Smart Watch
    0.05,  # Printer
    0.08,  # Backpack
    0.06,  # Office Chair
    0.05,  # Desk Lamp
    0.05,  # Coffee Maker
    0.05,  # Blender
    0.05   # Running Shoes
]

selected_products = np.random.choice(
    products,
    size=n,
    p=product_probabilities
)

# -----------------------------
# Date information
# -----------------------------

dates = pd.date_range(
    start="2024-01-01",
    end="2025-12-31",
    periods=n
)

dates = pd.to_datetime(dates)

# -----------------------------
# Quantity, price and discount
# -----------------------------

quantities = np.random.randint(1, 6, size=n)

unit_prices = [
    prices[product] * np.random.uniform(0.90, 1.10)
    for product in selected_products
]

discounts = np.random.choice(
    [0, 0.05, 0.10, 0.15, 0.20],
    size=n,
    p=[0.25, 0.25, 0.25, 0.15, 0.10]
)

# Sales calculation
sales = quantities * np.array(unit_prices) * (1 - discounts)

# Cost and profit
cost = sales * np.random.uniform(0.60, 0.85, size=n)
profit = sales - cost

# -----------------------------
# Create DataFrame
# -----------------------------

df = pd.DataFrame({
    "Order_ID": [f"ORD{str(i).zfill(5)}" for i in range(1, n + 1)],
    "Order_Date": dates,
    "Customer_ID": np.random.choice(customer_ids, size=n),
    "Gender": genders,
    "Age": ages,
    "Age_Group": age_groups,
    "Product": selected_products,
    "Category": [categories[p] for p in selected_products],
    "Quantity": quantities,
    "Unit_Price": unit_prices,
    "Discount": discounts,
    "Sales": sales,
    "Profit": profit
})

# Save dataset
df.to_csv("retail_sales_task1.csv", index=False)

print("Dataset created successfully!")
print("File saved as: retail_sales_task1.csv")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
```

```python
# Create Month column
df["Month"] = df["Order_Date"].dt.to_period("M").astype(str)

monthly_sales = df.groupby("Month")["Sales"].sum().reset_index()

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x="Month",
    y="Sales",
    marker="o"
)

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
df["Quarter"] = df["Order_Date"].dt.to_period("Q").astype(str)

quarterly_sales = df.groupby("Quarter")["Sales"].sum().reset_index()

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=quarterly_sales,
    x="Quarter",
    y="Sales",
    marker="o"
)

plt.title("Quarterly Sales Trend", fontsize=16)
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
age_sales = df.groupby("Age_Group")["Sales"].sum().reset_index()

age_order = ["18-25", "26-35", "36-45", "46-55", "56-65"]

age_sales["Age_Group"] = pd.Categorical(
    age_sales["Age_Group"],
    categories=age_order,
    ordered=True
)

age_sales = age_sales.sort_values("Age_Group")

plt.figure(figsize=(10, 6))

sns.barplot(
    data=age_sales,
    x="Age_Group",
    y="Sales"
)

plt.title("Sales by Customer Age Group", fontsize=16)
plt.xlabel("Age Group")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
gender_sales = df.groupby("Gender")["Sales"].sum().reset_index()

plt.figure(figsize=(8, 5))

sns.barplot(
    data=gender_sales,
    x="Gender",
    y="Sales"
)

plt.title("Sales by Gender", fontsize=16)
plt.xlabel("Gender")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
gender_count = df["Gender"].value_counts().reset_index()

gender_count.columns = ["Gender", "Number_of_Customers"]

plt.figure(figsize=(8, 5))

sns.barplot(
    data=gender_count,
    x="Gender",
    y="Number_of_Customers"
)

plt.title("Customer Distribution by Gender", fontsize=16)
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```
